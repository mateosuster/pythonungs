## Ejercicio 3: Módulo propio + `numpy`

### 3.1 Crear e importar un módulo propio
En esta misma notebook, cree un archivo `finanzas.py` con las siguientes funciones:

- `vf(VP, i, n)`: valor futuro (como en Ejercicio 1)
- `vp(VF, i, n)`: valor presente $VP = VF / (1+i)^n$
- `tasa_efectiva_anual(TNA, m)`: $TEA = (1 + \frac{TNA}{m})^m - 1$

Cada función debe tener docstring completo.

Luego, en la notebook, importe el módulo y pruebe las tres funciones.

````python
# --- EJERCICIO 3.1: CREAR MODULO PROPIO ---
# TODO: Escribir el archivo finanzas.py
with open('finanzas.py', 'w', encoding='utf-8') as f:
    f.write('''def vf(VP, i, n):
    """Calcula el valor futuro.

    Args:
        VP (float): Valor presente
        i (float): Tasa por período
        n (int): Número de períodos

    Returns:
        float: Valor futuro
    """
    return VP * (1 + i) ** n


def vp(VF, i, n):
    """Calcula el valor presente.

    Args:
        VF (float): Valor futuro
        i (float): Tasa por período
        n (int): Número de períodos

    Returns:
        float: Valor presente
    """
    return VF / (1 + i) ** n


def tasa_efectiva_anual(TNA, m):
    """Calcula la Tasa Efectiva Anual (TEA).

    Args:
        TNA (float): Tasa Nominal Anual
        m (int): Frecuencia de capitalización por año

    Returns:
        float: TEA
    """
    return (1 + TNA / m) ** m - 1
''')

print("Módulo finanzas.py creado")
````
