## Ejercicio 1: Fundamentos de funciones e `import`

### 1.1 `return` vs. `print`

`return` **devuelve** un valor que queda disponible para seguir operando (guardarlo en una variable, usarlo en otra cuenta, etc.). `print` solo **muestra** texto por pantalla y no devuelve nada (en realidad devuelve `None`). Por eso la consigna pide `return`: si la función hiciera `print`, `vf = calcular_valor_futuro(...)` guardaría `None` y no podríamos seguir calculando con ese resultado.

```python
def calcular_valor_futuro(VP, i, n):
    """Calcula el valor futuro de un capital.

    Args:
        VP (float): valor presente.
        i (float): tasa de interes por periodo (0.05 = 5%).
        n (int): cantidad de periodos.

    Returns:
        float: el valor futuro VF = VP * (1 + i)**n.
    """
    return VP * (1 + i) ** n


# Prueba: guardamos el resultado en una variable
vf = calcular_valor_futuro(10_000, 0.05, 10)
print(f"VF = {vf:.2f}")
```

### 1.2 Parámetros por defecto y argumentos *keyword*

Al redefinir la función con `i=0.05` y `n=12`, puede llamarse con 1, 2 o 3 argumentos: los que no se pasan toman su valor por defecto, y siempre se puede "saltear" un parámetro pasándolo por nombre (`n=6`).

```python
def calcular_valor_futuro(VP, i=0.05, n=12):
    """Calcula el valor futuro de un capital (i y n con valores por defecto)."""
    return VP * (1 + i) ** n


print(calcular_valor_futuro(10_000))                  # usa i=0.05 y n=12 por defecto
print(calcular_valor_futuro(10_000, 0.10))            # n sigue siendo 12
print(calcular_valor_futuro(10_000, n=6))            # i por defecto, n=6 por nombre
print(calcular_valor_futuro(VP=5_000, i=0.03, n=24))  # todos por nombre (keyword)
```

### 1.3 Tres formas de importar `math`

| Forma | Cómo queda el código | ¿Cuándo conviene? |
|---|---|---|
| `import math` | `math.log(x)` | La más explícita: siempre se ve de qué módulo viene cada función. Ideal cuando hay muchas cosas en juego |
| `from math import log, exp` | `log(x)` | Cuando se usan unas pocas funciones muchas veces: el código queda más corto |
| `import math as mt` | `mt.sqrt(x)` | Alias breve, típico para librerías de nombre largo (`import numpy as np`) |

Un cuidado de la segunda forma: si el programa ya tenía una función propia llamada `log`, el `import` la pisa (colisión de nombres). La primera y la tercera la evitan.

```python
VP = 10_000
VF = 16_288.95

# a) importar el modulo completo
import math
print(math.log(VF / VP))

# b) importar funciones puntuales
from math import log, exp
print(log(VF / VP), exp(1))

# c) importar con alias
import math as mt
print(mt.sqrt(VF))
```

## Ejercicio 2: Scope, módulos nativos y `random`

### 2.1 Variables locales vs. globales

Código original: imprime `Dentro: x = 20` pero `Afuera: x = 10`. La asignación `x = 20` dentro de `modificar()` **crea una variable local**: es *otra* `x` que solo existe mientras corre la función, que "tapa" (shadowing) a la global dentro de la función y desaparece al terminar. La `x` global nunca se toca.

Con `global x` se le indica a Python que, dentro de esa función, `x` refiere a la variable **global**, por lo que la asignación sí la modifica.

```python
# --- Codigo original ---
x = 10

def modificar():
    x = 20                       # crea una variable LOCAL llamada x
    print(f"Dentro: x = {x}")

modificar()
print(f"Afuera: x = {x}")
```

```python
# --- Version con `global` ---
x = 10

def modificar():
    global x                     # ahora x refiere a la global
    x = 20
    print(f"Dentro: x = {x}")

modificar()
print(f"Afuera: x = {x}")
```

**¿Es buena práctica?** En general, no. Una función que modifica variables globales produce efectos que no se ven en su firma: el estado entra "por la puerta de atrás". Eso vuelve el código más difícil de leer, testear y reutilizar, y es una fuente clásica de bugs ("¿quién me cambió esta variable?"). Salvo casos muy justificados, conviene que las funciones se comuniquen con el resto del programa por **parámetros y `return`**, y dejar el estado global lo más inerte posible.

### 2.2 Simulación estocástica con `random`

Se fija la semilla (`random.seed`) para que los números de esta resolución sean **reproducibles**; comentando esa línea, cada corrida da resultados distintos (que es lo natural en una simulación).

```python
import random

random.seed(42)   # semilla: hace reproducible la simulacion (opcional)

def simular_tiradas(n):
    """Simula n tiradas de un dado de 6 caras.

    Args:
        n (int): cantidad de tiradas a simular.

    Returns:
        tuple: (promedio, minimo, maximo, frecuencia_6), donde frecuencia_6
            es la proporcion de tiradas que salieron 6.
    """
    suma = 0
    minimo = None
    maximo = None
    contador_seis = 0
    t = 0
    while t < n:                   # el TP pide recorrer con while
        dado = random.randint(1, 6)
        suma += dado
        if minimo is None or dado < minimo:
            minimo = dado
        if maximo is None or dado > maximo:
            maximo = dado
        if dado == 6:
            contador_seis += 1
        t += 1
    return (suma / n, minimo, maximo, contador_seis / n)
```

```python
# Probamos con distintos tamanos de muestra, recorriendolos con while
tamanos = [10, 1_000, 100_000]
j = 0
while j < len(tamanos):
    n = tamanos[j]
    prom, mn, mx, f6 = simular_tiradas(n)
    print(f"n={n:>7,}: promedio={prom:.3f}, minimo={mn}, maximo={mx}, frecuencia_6={f6:.4f}")
    j += 1

print()
print("Valores teoricos: promedio = (1+2+3+4+5+6)/6 =", (1 + 2 + 3 + 4 + 5 + 6) / 6)
print("                  frecuencia de 6 = 1/6 =", round(1 / 6, 4))
```

**Convergencia (ley de los grandes números):** con `n` pequeño, el promedio puede caer lejos del valor teórico (3,5) y la frecuencia de seises oscila mucho. A medida que `n` crece, ambos estadísticos se "pegan" a sus valores esperados: el promedio tiende a **3,5** y la frecuencia de 6 a **1/6 ≈ 0,1667**. El mínimo y el máximo, en cambio, llegan rápido a 1 y 6 y ahí se quedan: no promedian nada, solo registran valores extremos.

## Ejercicio 3: Módulo propio + `numpy`

### 3.1 Crear e importar un módulo propio

Con la celda mágica `%%writefile` creamos el archivo `finanzas.py` desde la propia notebook (equivalente a escribirlo a mano o con `open('finanzas.py', 'w')` como en la plantilla del TP). Después se importa como a cualquier módulo.

**Atención (para Colab/Jupyter):** si se edita `finanzas.py` y se vuelve a correr el `import`, Python puede seguir usando la versión vieja, porque el módulo ya quedó cargado en memoria. Hay que reiniciar el entorno o recargarlo con `importlib.reload(finanzas)`.

```python
%%writefile finanzas.py
def vf(VP, i, n):
    """Calcula el valor futuro de un capital.

    Args:
        VP (float): valor presente.
        i (float): tasa de interes por periodo.
        n (int): cantidad de periodos.

    Returns:
        float: valor futuro VF = VP * (1 + i)**n.
    """
    return VP * (1 + i) ** n


def vp(VF, i, n):
    """Calcula el valor presente de un capital futuro.

    Args:
        VF (float): valor futuro.
        i (float): tasa de interes por periodo.
        n (int): cantidad de periodos.

    Returns:
        float: valor presente VP = VF / (1 + i)**n.
    """
    return VF / (1 + i) ** n


def tasa_efectiva_anual(TNA, m):
    """Convierte una TNA en Tasa Efectiva Anual (TEA).

    Args:
        TNA (float): tasa nominal anual (decimal, ej. 0.40).
        m (int): frecuencia de capitalizacion por ano (12 = mensual).

    Returns:
        float: TEA = (1 + TNA/m)**m - 1.
    """
    return (1 + TNA / m) ** m - 1
```

```python
import finanzas

print(f"VF(10000, 0.05, 10)    = {finanzas.vf(10_000, 0.05, 10):.2f}")
print(f"VP(16288.95, 0.05, 10) = {finanzas.vp(16_288.95, 0.05, 10):.2f}")
print(f"TEA(TNA=40%, m=12)     = {finanzas.tasa_efectiva_anual(0.40, 12):.4%}")
```

Las funciones son consistentes entre sí: el VP de $16.288,95 (que era el VF de $10.000 al 5% en 10 períodos) vuelve a dar $10.000. Y la TNA del 40% con capitalización mensual equivale a una TEA del ≈48,2%: capitalizar más seguido que lo que la tasa "nombra" aumenta la tasa efectiva.

### 3.2 Arreglos y funciones vectorizadas con `numpy`

La gracia de `numpy` es operar sobre **todo el array de una vez**, sin bucles explícitos: `(1 + i) ** n_array` devuelve un array con la potencia calculada elemento a elemento (*broadcasting*). Además de más corto, es mucho más rápido que procesar los elementos de a uno 

```python
import numpy as np

def vf_vectorizado(VP, i, n_array):
    """Calcula el VF para cada n de n_array, de forma vectorizada.

    Args:
        VP (float): valor presente.
        i (float): tasa de interes por periodo.
        n_array (array o lista de int): periodos a calcular.

    Returns:
        np.ndarray: VF para cada elemento de n_array.
    """
    n_array = np.asarray(n_array)     # convierte listas a array (si ya lo es, no hace nada)
    return VP * (1 + i) ** n_array


n1 = [1, 2, 3, 5, 10, 20]
n2 = np.arange(1, 13)               # meses 1 a 12

print(vf_vectorizado(10_000, 0.05, n1))
print(vf_vectorizado(10_000, 0.05, n2))
```
